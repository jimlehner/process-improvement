# xmr_chart function demo

```python
# Import libraries
from process_improvement.charts.xmr_charts import xmr_chart
from process_improvement.charts.utils import XmRChartConfig

from process_improvement.data_loader import list_example_files, load_example_data

%matplotlib inline
```

## Get data

```python
# List example files
list_example_files()
```

```python
# Create dataframe
df = load_example_data("OP200_weekly_first_pass_yield")

df.head()
```

## XmRChartConfiguration

```python
# XmR chart configuration
config = XmRChartConfig(
    tickinterval=1,
    round_value=1,
    show_xticks=False,
    show_yticks=False,
    show_limit_values=False
)
```

## XmR chart of first-pass yield

```python
# Create XmR chart
results = xmr_chart(
    df, 
    'First-Pass Yield (%)', 
    'Week', 
    config=config
)
```

## Show process statistics

```python
# Show statistics
results.stats_df
```

## Show just the X chart portion of the XmR chart

```python
# Show just the X chart
x_ax = results.axes[0]
# Hide the mR axis
for ax in x_ax.figure.axes:
    if ax != x_ax:
        ax.set_visible(False)

x_ax.figure
```
